```python
#function to take input value for weights and biases
def initialize_parameters(layer_dims):
  #layer_dims- input, hidden layer,hiden,output
  parameters = {}
  L = len(layer_dims)
  for l in range(1, L):
    parameters['W' + str(l)] = np.ones((layer_dims[l-1], layer_dims[l])) * 0.1 # w1=(shape of the matrics(value=0.1)) weight is for every arrow
    parameters['b' + str(l)] = np.zeros((layer_dims[l], 1))#bias is foer every neuron current node,1 to make the shape 2D
  return parameters
```

```python
#to caluculte the output of any given neuron
def linear_forward(A_prev, W, b):

  Z = np.dot(W.T, A_prev) + b

  return Z
```

```python
def L_layer_forward(X, parameters):
  A = X
  L = len(parameters) // 2
  for l in range(1, L+1):
    A_prev = A
    Wl = parameters["W" + str(l)]
    bl = parameters["b" + str(l)]
   # print("A"+str(l-1)+": ", A_prev)
   # print("W"+str(l)+": ", Wl)
    #print("b"+str(l)+": ", bl)
    A = linear_forward(A_prev, Wl, bl)
    #print("A"+str(l)+": ", A)
    #print("***"*20)
  return A, A_prev
```

```python
def update_parameters(parameters, y, y_hat, A1, X):

  # 1. Save the ORIGINAL W2 weights for the chain rule
  old_W2_0 = parameters['W2'][0][0]
  old_W2_1 = parameters['W2'][1][0]

  # 2. Update Output Layer (Layer 2)
  parameters['W2'][0][0] = parameters['W2'][0][0] + (0.001 * 2 * (y - y_hat) * A1[0][0])
  parameters['W2'][1][0] = parameters['W2'][1][0] + (0.001 * 2 * (y - y_hat) * A1[1][0])

  # BUG 1 FIXED: Using b2 here instead of W2
  parameters['b2'][0][0] = parameters['b2'][0][0] + (0.001 * 2 * (y - y_hat))

  # 3. Update Hidden Layer (Layer 1)
  # BUG 2 FIXED: Using old_W2 for the chain rule
  parameters['W1'][0][0] = parameters['W1'][0][0] + (0.001 * 2 * (y - y_hat) * old_W2_0 * X[0][0])
  parameters['W1'][0][1] = parameters['W1'][0][1] + (0.001 * 2 * (y - y_hat) * old_W2_0 * X[1][0])
  parameters['b1'][0][0] = parameters['b1'][0][0] + (0.001 * 2 * (y - y_hat) * old_W2_0)

  parameters['W1'][1][0] = parameters['W1'][1][0] + (0.001 * 2 * (y - y_hat) * old_W2_1 * X[0][0])
  parameters['W1'][1][1] = parameters['W1'][1][1] + (0.001 * 2 * (y - y_hat) * old_W2_1 * X[1][0])
  parameters['b1'][1][0] = parameters['b1'][1][0] + (0.001 * 2 * (y - y_hat) * old_W2_1)
```

```python
parameters = initialize_parameters([2,2,1])
epochs = 75

for i in range(epochs):

    Loss = []

    for j in range(df.shape[0]):

        X = df[['cgpa','profile_score']].values[j].reshape(2,1)
        y = df[['lpa']].values[j][0]

        y_hat, A1 = L_layer_forward(X,parameters)
        y_hat = y_hat[0][0]

        update_parameters(parameters,y,y_hat,A1,X)

        Loss.append((y - y_hat)**2)

    print('Epoch - ',i+1,'Loss - ',np.array(Loss).mean())
```
